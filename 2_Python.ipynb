## Data types and expressions
### Integers (`int`)
Integers in Python have the same meaning as in mathematics: these are numbers with no values after the decimal point and can be positive, negative or zero.  For example: 10, 23 and -18 are all integers.  In contrast: 1.5, -0.2 and $\sqrt2$ are not integers, since they have values after the decimal point.

Integers can be manipulated by operators.  The **addition (+) and subtraction (-) operators** will be familiar to you from basic maths.

Calculations can be performed in Jupyter Notebook cells:

```python
1 + 2
```

```python
3 - 5
```

Python can also perform multiplication.  The multiplication symbol used by Python is not the same as used in standard maths textbooks, for it is an asterisk (*). 

```python
5 * 5
```

Furthermore, it is possible in Python to raise a number to a given power. This is denoted by a **double asterisk operator (\**)**. 

```python
5 ** 2
```

### Floats (`float`)
Floats - or to use the full term: “floating point numbers" - are used by computers to store non-integer numbers.

Floats have two components: the **significand** and the **exponent**.  The former stores the significant numbers (which can be positive or negative), while the exponent defines the position of the decimal place.  For example, the value 0.5 has a significand of 5 and an exponent of -1.  

(In fact, integer values may be represented as floats, for example 1000 would have a significant of 1.0 and an exponent of 3, but storing integers in this way uses more of the available memory than simply storing as an `int` datatype.)

Having said all this, you generally won’t notice any difference when entering floats in the Jupyter Notebook as compared to integers.

```python
# A float: significand:1, exponent: -1
0.1
```

One noteworthy difference is that floats may be entered using scientific notation: 

```python
3.0E-10
```

Floating point numbers are particularly important with regards to division.  This is because most division operations (even those involving only integers) will return fractional values and thus a `float`.  Indeed, division operations in Python always return the `float` type (even if the value returned is not fractional).

To divide in Python, use the forward slash (/) character:

```python
1 / 3
```

<div class="alert alert-info">

**A note on floats**

Many numbers cannot be reported with infinite precision using standard decimal notation.  For example, one-third is represented as 0.33333 recurring, but since we cannot store an infinite number of 3s in memory, the stored value will only an approximation to one-third.

In a similar vein, computers work in binary (base 2) and there are seemingly simple decimal values that cannot be stored with complete precision in binary.  For example, the value 0.1 in binary is represented as 1.100110011... × 2<sup>-4</sup> which equates to  0.10000000000000000555 to 20 decimal places in decimal.  Consequently Floats are accurate to around 17 decimal places in Python, but look out for these slight inaccuracies when performing calculations.
    

    # An example of floating-point imprecision
    0.1 ** 2
    0.010000000000000002

Later in the course you will learn how to round number to a specified number of decimal places.  Rounding the above result to 5 decimal places will equate to: 0.01000.
<hr>    
<b>Warning: these tiny discrepancies do not cause a problem in almost all situations, but do be careful when checking if two results exactly equal one another when working with floats.</b>
</div>

### Compound expressions
All the expressions so far listed involve one operator and one value, but a single expression may contain many values and expressions.  Here is a simple example involving only integers:

```python
5 + 5 * 3
```

If one performed the above calculation in a left-to-right order, the answer would be 30 and not 20 (i.e. 5 + 5 = 10 and then 10 * 3 = 30).  

However, Python follows **orders of precedence** and will perform the multiplication before the addition (which are the formal rules in mathematics).  Thus, 5 * 3 = 15.  Then add 5, which makes 20. You could of course learn the rules of precedence and write code accordingly, programmers tend not to do this.  

The way around this extra complexity is to use parentheses in your code.  The expression inside the parentheses is evaluated first and then this is passed to values outside the parentheses.  Python also allows nested parentheses (brackets within brackets), which are useful in complex expressions.  Don’t worry about using multiple parentheses in your code; in fact, their inclusion should help make your code easier to read.

Consequently, you could re-write the above expression as:

```python
(5 + 5) * 3
```

### Strings (`str`)

These store (Unicode) characters, whether that is a letter, number or a symbol of some kind.  Essentially, they constitute a “string” of characters – although a solitary single character is also classified as a string.  

To create a string you will need to enclose your text within either single or double quotation marks.  Double and single quotation marks have the same meaning in Python and you are free to use either, but the opening and closing quotes need to be of the same type.

```python
# Create a string with single quotation marks
'abcde12345'
```

```python
# Create a string with double quotation marks
"abcde12345"
```

Placing numbers between quotation marks will create a string, not a numerical datatype.

```python
'12345'
```

Although you may think of plus (+) as to be used solely with numbers, in Python it can also be used to concatenate (i.e. join) strings.

```python
'Love' + 'Marriage'
```

In a similar fashion, the multiplication operator can be used on strings. Multiplying a string by an integer causes the string to be repeated.

```python
'Go forth ' * 5
```

#### String indexing and slicing

Indexing and slicing data are concepts in Python that come up again and again in different contexts, so we may as well introduce these ideas now, early in the course, with respect to strings. 

Strings are **ordered** characters and so we can assign each character within a string a number, **starting at 0** and incrementing by 1 as we move across the string from left to right.  Using this indexing convention it is possible to retrieve single or multiple characters from a string as required.

The indexing syntax involves placing square brackets after the string.  Inside these brackets we can enter the position (remember indexing starts at 0!) of the character to be returned.

So, to retrieve the first element of a string:

```python
'abcdefg'[0]
```

And, to retrieve the second element of a string:

```python
'abcdefg'[1]
```

There is also an "alternative" indexing system that starts on the right-hand, which is assigned the value -1.  As we move move from right to left, the value of the index decreases by 1.

To retrieve the last element of a string:

```python
'abcdefg'[-1]
```

To retrieve the fourth-to-last element of a string (notice also that double-quotes may be used interchangeably with single quotes when declaring strings):

```python
"abcdefg"[-4]
```

The figure below illustrates this indexing nomenclature.

<img src='course_images/slicing2.svg' title='Slicing'/>

It is possible to retrieve multiple characters from a string using **slicing**.  In slicing, the **colon operator (`:`)** is used to specify contiguous positions.

In the example below, the number before the colon references the start index of the slice.  The number after the colon references the end position - but **the slice terminates immediately before the end index**.

So the example below retrieves the first character of a string and extends up to, but not including, the fourth character of a string:

```python
'abcdefg'[0:3]
```

In fact, there is shorthand notation to specify starting the slice at the first position, and that is to enter nothing before the colon.  The example below is therefore equivalent to the example above.

```python
'abcdefg'[:3]
```

Likewise, to retrieve the 'cde' letters from towards the middle of the string we would use the code:

```python
'abcdefg'[2:5]
```

The code below retrieves a slice from the fourth character of the string to the penultimate position:

```python
'abcdefg'[3:6]
```

How then would we retrieve the last position of a string?  Well, to do this we leave the term after the colon blank:

```python
'abcdefg'[3:]
```

(In fact, it is allowed in Python to define a slice that extends outside of the actual string.  Python will return all the available characters that fall within the slice.)

```python
# This works, but it is not very elegant
'abcdefg'[0:1000000]
```

It is also possible to move across a string by a constant step size.  To do this, simply add a second colon and place the increment value after that.  In the example below, every second letter is returned from the start to fifth position of the string.

```python
'abcdefg'[0:6:2]
```

The example below will return every second character from the entire string (note the juxtaposed colons that specify that an entire string should be returned).

```python
'abcdefg'[::2]
```

### Booleans (`bool`)
The Boolean datatype has only two values: `True` or `False`.  Boolean values are generated when performing logical evaluations using **comparison operators**.  This term may sound quite formal, but you are no doubt already familiar with many of these operators from school-level maths.

| Comparison Operator | Description |
| --------------------| ----------- |
|          ==         | If the values of two operands are **equal**, then the condition is `True` | 
|          !=         | If values of two operands are **not equal**, then the condition is `True` |
|           >         | If the value of left operand is **greater than** the value of right operand, then the condition is `True`   |
|           <         | If the value of left operand is **less than** the value of right operand, then the condition is `True`  |
|           >=        | If the value of left operand is **greater than or equal to** the value of right operand, then the condition is `True`  |
|          <=         | If the value of left operand is **less than or equal to** the value of right operand, then the condition is `True`  |

For all of the above, if the condition is not met then the returned value is `False`.

```python
1 == 1
```

```python
1 == 0
```

```python
2 > 5
```

Please note that `True` or `False` values are case sensitive.  Not following this syntax will raise an error.

```python
True
```

### Lists
Lists constitute a **mutable ordered sequence** of any type of element.  The code below shows how to create a list.

```python
a = [1, 2, 3]
print(a)
print(type(a))
```

You may reference a specific element in a list using an indexing system.  The first index has a value 0, then next index has the value 1 and so on.  Place the index value between square brackets to access the item at that position.

### Dictionaries

Dictionaries (`dict`) are mutable structures that store data in what is known as **“key/value”** pairs.
```python
range(3)
```

The returned value will show that a range has been created and also display the **start** and **stop** values.  If we pass this range to a list, you will see that the values of the range are 0, 1, and 2 but not 3:

### Conditionals
A central concept in computing is the capability to evaluate a particular value, or set of values, and then make a “decision” based on this input.  In Python, such conditional statements are created using the `if` **keyword**.  The example below is a simple conditional expression: 

```python
if 1==1:
    print('One is equal to one')
```

<div class="alert alert-info">

<b>Python Keywords</b>

`if` is not a Python function, which is why round brackets are not required immediately afterwards.  `if` is actually a Python **keyword**.  Python has a set of keywords that are reserved words which cannot be used as variable names or function names, etc.  Other keywords include `True` and `False` which we encountered previously.  
</div>

You can think of the `if` keyword much the same as the English language equivalent, for it asks the question: “is the following true?”  

The above example asks whether one is equal to one – which of course it is.  The statement evaluating to `True` causes the code on the next line to be executed, whereas conversely `False` would cause the next line to be ignored.  

Notice at this point that the indentation scheme: the conditional expression ends with a colon and the print statement that may be performed depending of the outcome of the `if` statement, is indented by 4 spaces.

Let’s illustrate the structure once again, but this time with a one-alternative conditional:

```python
if 2 + 2 == 5:
    print('Two plus two is five')
else:
    print('Two plus two is not equal to five')
```

In this example, the conditional expression evaluates to `False`.  As before, this will cause the statement after the `if` keyword to be ignored, but the statement after the `else` keyword will be printed to the screen.  This should make intuitive sense: if true do this, else do that.  On many occasions, clear Python can almost be read as instructions written in the English language.

### Loops
Loops are a central concept in many programming languages.  They cause the same block of code to be repeated (i.e. looping) until a fixed number of repetitions has been achieved, or some other criterion is has been satisfied.  There are many different types of loops in Python, each one suited to slightly different situations.

#### While Loops
With much of the Python code it is possible for a beginner to make a reasonable guess at what a statement does simply from the meaning of the words in English.  The `while` loop is such example, and in simple terms the statement means: keep looping **while** this is **true**.  

For example:

```python
x = 1 
while x < 5:
    print(x)
    x = x + 1 
```

Is this code, `x` is assigned the value 1.  We then proceed to the loop which will be carried out for so long as x is less than 5.  At the end of this line of code we have a colon followed by the four-space line indentation on the next line.  This indented code will be performed for each loop.  The `print` statement will display the value of x on the screen, and then the value of x is incremented by 1.  This will cause the values 1 to 4 (not 5, since x need to be less than 5) to be printed to the screen.  

It is worth pointing out that if the increment statement had not been included, x would have the value of 1 and the loop would be repeated forever; we would have created an infinite loop.  Watch out for these when writing your code!  If a program or operation has not terminated after a much longer time than expected, re-check that you have not inadvertently created an infinite loop. 

![while_loop_schematic](course_images/while_loop_schematic.png)

You may append an else statement to your while loop which will be executed when the loop condition evaluates to False:

```python
x = 1 

while x < 5:
    print(x)
    x = x + 1
else:
    print("Not less than 5")
```
